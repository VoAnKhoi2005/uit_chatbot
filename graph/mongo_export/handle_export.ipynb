```python
import json

# ---------- Concepts ----------
with open("KB_UIT.concepts.json", "r", encoding="utf-8") as f:
    concepts = json.load(f)

concept_names = [c["name"] for c in concepts]

with open("concepts.txt", "w", encoding="utf-8") as f:
    for name in concept_names:
        f.write(name + "\n")

# ---------- Relations ----------
with open("KB_UIT.relations.json", "r", encoding="utf-8") as f:
    relations = json.load(f)

relation_names = [r["name"] for r in relations]

with open("relations.txt", "w", encoding="utf-8") as f:
    for name in relation_names:
        f.write(name + "\n")

# ---------- Triplets ----------
with open("KB_UIT.triplets.json", "r", encoding="utf-8") as f:
    triplets = json.load(f)

triplet_list = [[t["subject_name"], t["relation_name"], t["object_name"]] for t in triplets]

with open("triplets.txt", "w", encoding="utf-8") as f:
    for triplet in triplet_list:
        f.write(f"{triplet[0]} -> {triplet[1]} -> {triplet[2]}\n")

print("3 text files generated: concepts.txt, relations.txt, triplets.txt")
```
