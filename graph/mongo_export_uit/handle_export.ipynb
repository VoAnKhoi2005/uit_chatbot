```python
from graph.src.db import init_sqlite

conn, cursor = init_sqlite(r"E:\Github\uit_chatbot\graph\jupyter\uit_law.db")
```

```python
from graph.src.db import extract_from_sqlite, init_sqlite
import json

# ---------- Concepts ----------
with open("v3/KB_UIT.concepts.json", "r", encoding="utf-8") as f:
    concepts = json.load(f)

concept_names = [c["name"] for c in concepts]

with open("v3/concepts.txt", "w", encoding="utf-8") as f:
    for name in concept_names:
        f.write(name + "\n")

# ---------- Relations ----------
with open("v3/KB_UIT.relations.json", "r", encoding="utf-8") as f:
    relations = json.load(f)

relation_names = [r["name"] for r in relations]

with open("v3/relations.txt", "w", encoding="utf-8") as f:
    for name in relation_names:
        f.write(name + "\n")

# ---------- Triplets ----------
with open("v3/KB_UIT.triplets.json", "r", encoding="utf-8") as f:
    triplets = json.load(f)

triplet_list = [[t["subject_name"], t["relation_name"], t["object_name"]] for t in triplets]

with open("v3/triplets.txt", "w", encoding="utf-8") as f:
    for triplet in triplet_list:
        f.write(f"{triplet[0]} -> {triplet[1]} -> {triplet[2]}\n")

# Make sure triplet_list includes document_id
triplet_list = [
    [t["subject_name"], t["relation_name"], t["object_name"], t["document_id"]]
    for t in triplets
]

# Keep track of which documents we've already written
written_docs = set()

with open("v3/triplets.txt", "w", encoding="utf-8") as f:
    for triplet in triplet_list:
        subj, rel, obj, doc_id = triplet

        # If this document hasn't been processed yet, write sentence first
        if doc_id not in written_docs:
            rows = extract_from_sqlite(cursor, doc_id)
            if rows:
                sentence = rows[0].get("content", "No content found")  # content column
                f.write(f"Original sentence: {sentence}\n")
            else:
                f.write("Original sentence not found\n")
            written_docs.add(doc_id)

        # Write the triplet
        f.write(f"{subj} -> {rel} -> {obj}\n")

        # If this is the last triplet for this document, add a newline
        # We'll check the next triplet's doc_id
        current_index = triplet_list.index(triplet)
        next_index = current_index + 1
        if next_index >= len(triplet_list) or triplet_list[next_index][3] != doc_id:
            f.write("\n")


print("3 text files generated: concepts.txt, relations.txt, triplets.txt")
```
